```python
from rbc_dataloader import RBC_Dataloader

dataloader = RBC_Dataloader(
    datadir=Label_Free_Path,
    transform=None
)

# Try accessing by index instead
if len(dataloader) > 0:
    sample = dataloader[0]
    print(f"Sample keys: {sample.keys()}")
    print(f"Image shape: {sample['image'].shape}")
    print(f"Label: {sample['label']}")
    print(f"Classifier: {sample['classifier_name']}")
else:
    print("Dataloader is empty!")
```

```python
for i in range(len(dataloader)):
    sample = dataloader[i]
    print(f"Sample {i} keys: {sample.keys()}")
    print(f"Image shape: {sample['image'].shape}")
    print(f"Label: {sample['label']}")
    print(f"Classifier: {sample['classifier_name']}")
```
